```python
# these are the calibration curves that Simone developed
# for the soil groups at our sites. Calibrations is the
# fitted parameters for the equation, and calgroups assigns
# each sensor to a calibration group

calibrations = {'sentekDefault':[0.1957,0.404,0.0285],
               'group1': [0.00878, 0.952, 0.527],
               'group2': [0.21336, 0.39122, 0],
               'group3': [0.12, 0.53, 0],
               'group4': [0.1474, 0.47659, 0]}

calgroups = {'WP1':{0.3:'group4',0.5:'group4',0.7:'group4',
                   1.0:'group4',1.4:'group4',
                   1.7:'group4',2.0:'group4',
                   2.4:'group4'},
            'WP2':{0.25:'group4',0.45:'group4',
                  0.65:'group4',0.95:'group4',
                  1.25:'group4',1.55:'group4',
                   1.7:'group4',1.85:'group4',
                  2.05:'group4',2.4:'group3'},
            'GN':{0.2:'group4',0.4:'group4',
                 0.6:'group4',1.0:'group4',
                 1.5:'group4',2.0:'group4',
                 2.5:'group3',2.8:'group3',
                 3.0:'group3',3.2:'group3'},
             'GN2':{0.2:'group4',0.4:'group4',
                 0.6:'group4',1.0:'group4',
                 1.5:'group4',2.0:'group4',
                 2.5:'group3',2.8:'group3',
                 3.0:'group3',3.2:'group3'},
             'GN1':{0.2:'group4',0.4:'group4',
                 0.6:'group4',1.0:'group4',
                 1.5:'group4',2.0:'group4',
                 2.5:'group3',2.8:'group3',
                 3.0:'group3',3.2:'group3'},
             'GN3':{0.2:'group4',0.4:'group4',
                 0.6:'group4',1.0:'group4',
                 1.5:'group4',2.0:'group4',
                 2.5:'group3',2.8:'group3',
                 3.0:'group3',3.2:'group3'},
             'GN4':{0.2:'group4',0.4:'group4',
                 0.6:'group4',1.0:'group4',
                 1.5:'group4',2.0:'group4',
                 2.5:'group3',2.8:'group3',
                 3.0:'group3',3.2:'group3'},
            'GG1':{0.25:'group3',0.45:'group3',
                  0.65:'group3',0.85:'group3',
                  1.05:'group2',1.55:'group2',
                  1.95:'group2',2.55:'group3',
                  2.95:'group3',3.95:'group2',
                  4.95:'group2',5.95:'group2',
                  6.95:'group2',7.95:'group2',
                  8.55:'group2',9.05:'group2'},
            'DD':{0.25:'group1',0.45:'group1',
                 0.65:'group1',0.95:'group1',
                 1.45:'group1',1.95:'group1',
                 2.45:'group1',2.95:'group1',
                 3.95:'group1',4.95:'group2',
                 5.95:'group1',6.95:'group1',
                 7.35:'group1',7.95:'group1'}}
```

```python
# year to start a drift correction for each sensor
# these are depths when we think there may be a rift correction needed
# these sensors may need to be checked for rust or recalibrated.
# decisions were made via visual inspection and comparison to
# other hydrological datasets. These can be adjust here.

driftCorrectStart = {'WP1':{
                            2.0: 2022,
                            2.4: 2023,
                           },
                     'WP2':{
                            2.05: 2022,
                     },
                     'GG1':{0.65: 2024,
                            0.85: 2024,
                            1.55: 2024,
                            1.95: 2024,
                            1.95: 2024, 
                            2.55: 2024,
                            2.95: 2024, 
                            3.95: 2023, 
                            4.95: 2022,
                     },
                     'DD':{
                           4.95: 2024,
                           5.95: 2024,
                           6.95: 2024, 
                           7.95: 2023, 
                     }
                    }
# not dropping any sensors at this point
dropsensors = {'WP2':[],#[1.85],
              'GG1':[],#[9.05],
              'DD':[],}#[7.95]}

# irrimax naming conventions. for GN1-GN4, I named those attempting
# to process the downloaded files from the field.
afsystemsnames = {'DD':'065',
                 'GG1':'066',
                 'GN':'074',
                 'WP1':'063',
                 'WP2':'064',
                 'GN1':'S1',
                 'GN2':'S2',
                 'GN3':'S3',
                 'GN4':'S4'}
```

```python
# this block takes the input data files from irrimax saved in 
# ../Data/raw_sentek/ and processes them into final form
# and saves them to ../Data/temp_processing/

dataloc = '../Data/raw_sentek/'
files = {site:[f for f in os.listdir(dataloc) if f.endswith('csv') if afsystemsnames[site] in f][0]
        for site in sites}

for site in sites[:]:
    print(site)
    
    ## Step 0: read in raw sentek data
    original = pd.read_csv(dataloc+files[site])
    original['Date Time'] = pd.to_datetime(original['Date Time'])
    original.set_index('Date Time',inplace=True)
    for c in original.columns:
        original[c] = np.where(original[c]<0,np.nan,original[c])
    
    ## Step 1: apply Simone's calibration
    # reverse_processing from cooked sentek data to get raw
    # then, apply Simone's calibration curves
    
    processed = original.copy()
    caluse = calibrations['sentekDefault']
    if site=='DD':
        deplist = processed.columns[:-3]
    else:
        deplist = processed.columns
    for c in deplist:
        processed[c] = ((processed[c]-caluse[2])/caluse[0])**(1/caluse[1])
        processed[c] = caluse[0]*(processed[c]**caluse[1])+caluse[2]
        depth_m = float(c.split('(')[1][:-1])/100
        calgroup = calgroups[site][depth_m]
        calset = calibrations[calgroup]
        processed[c] = calset[0]*(processed[c]**calset[1])+calset[2]
        processed[c] = ((processed[c]-calset[2])/calset[0])**(1/calset[1])
        
        # remove WP1 data when water flooded sensors
        if site=='WP1':
            processed[c] = np.where((processed.index>pd.to_datetime('3-1-2025'))&
                                  (processed.index<pd.to_datetime('7-1-2025')),
                                  np.nan,processed[c])
            processed[c] = processed[c].ffill()
            
            # one sensor at WP1 shows a step change after the water intrusion event
            # subtract this step change
            if float(c.split('(')[1][:-1])<190:
                pdiff = processed[c].loc[pd.to_datetime('8-1-2025'):].min()-processed[c].loc[:pd.to_datetime('3-1-2025')].min()
                processed[c] = np.where(processed.index>=pd.to_datetime('7-1-2025'),
                                       processed[c]-pdiff,
                                       processed[c])
    processed.rename(columns={c:float(c.split('(')[1][:-1])/100 for
                                c in processed.columns},inplace=True)
    
    ## Step 2: apply other miscellaneous data corrections
    
    # get rid of extra columns in DD that don't have data
    if site=='DD':
        for col in processed.columns:
            if len(processed[col].dropna())<len(processed)/4:
                processed = processed.drop(columns=col)
    
    # grab the missing data (not on Irrimax) for WP2
    if site=='WP2': 
        gfill = pd.read_csv('../Data/other_inputs/WP2_Sentek_TwoYrs.csv',skiprows=3)
        gfill['Date Time'] = pd.to_datetime(gfill['Date Time'])
        gfill.set_index('Date Time',inplace=True)
        coluse = [c for c in gfill.columns if 'SWC_G' in c]
        gfill = gfill[coluse].rename(columns={col:float(col.split('_')[-1][:-2])/100
                                     for col in coluse})
        gfill.rename(columns={12.50:1.25},inplace=True)
        offset = gfill.loc['2-2-2023':'7-18-2023'].iloc[0]-processed.loc[:'2-2-2023'].iloc[-1]
        processed = pd.concat([processed,
                               gfill.loc['2-2-2023':'7-18-2023']-offset]
                             ).sort_index().resample('30min').mean()            

    if site=='DD': # sensor malfunction
        processed[6.95] = np.where(processed.index<pd.to_datetime('11-1-2023'),
                           np.nan,processed[6.95])
        processed[7.95] = np.where((processed.index>pd.to_datetime('2025-4-5'))&
                           (processed.index<pd.to_datetime('2025-9-12')),
                           np.nan,processed[7.95])
    if site in dropsensors.keys():
        processed = processed.drop(columns=dropsensors[site])
        
        
    ## Step 3: Drift correct sensors, as needed
    sm = processed.copy()
    
    if site not in ['GN1','GN3','GN4','GN','GN2']:
        for c in sm.columns[:]:
            if c in driftCorrectStart[site].keys():
                mindrift = sm[[c]]
                mindrift = mindrift[mindrift.index.year==driftCorrectStart[site][c]]
                minval = float(mindrift.min())
                minend = sm[sm.index.year==sm.index.year.max()][[c]]
                minvalend = float(minend.min())
                mindateend = minend[minend[c]==minvalend].index.values[0]
                mindate = mindrift[mindrift[c]==minval].index.values[0]
                run = len(sm.loc[mindate:mindateend])
                smcorrect = sm[[c]].copy()
                smsplit0 = smcorrect.loc[:mindate].iloc[:-1]
                smsplit1 = smcorrect.loc[mindate:]
                smsplit1 = smsplit1[c]-(minvalend-minval)/run*np.linspace(0,len(smsplit1),len(smsplit1))
                sm[c] = pd.concat([smsplit0,smsplit1])
                # make sure data remain positive
                if sm[c].min()<0:
                    sm[c] = sm[c]-sm[c].min()+.1
        
    
    ## Step 4: resample to daily by using the median daily measurement
    # opted for this over a 5am measurement because it's almost 
    # the same but a little smoother
    smD = sm.resample('D').median().ffill()
    processedD = processed.resample('D').median().ffill()
    
    ## Step 5: add daily quality flags
    for df in [smD,sm]:
        df['Quality_flags'] = 0
        if site == 'GG1':
            # mask first year for installation settling
            # 1 - installation settling
            df['Quality_flags'] = np.where(df.index<pd.to_datetime('4-1-2023'),
                                          1,df.Quality_flags)
            # mask second year for leaf litter blockage
            # 2 - litter cover, resolved by site maintenance
            df['Quality_flags'] = np.where((df.index>=pd.to_datetime('4-1-2023'))&
                                           (df.index<pd.to_datetime('4-1-2024')),
                                          2,df.Quality_flags)
        if site == 'DD':
            # mask first year for installation settling
            # 1 - installation settling
            df['Quality_flags'] = np.where(df.index<pd.to_datetime('1-1-2023'),
                                          1,df.Quality_flags)
            
    ## Step 5: export data to csv in ../Data/temp_processing/
    ff = '%.3f'
    sm.to_csv('../Data/temp_processing/vwc_processed_temp_'+site+'.csv',float_format = ff)
    smD.to_csv('../Data/temp_processing/vwc_processed_temp_'+site+'_daily.csv',float_format = ff)
    
    
    # Step 6: visualize to make sure everything looks good
    fig,axs = plt.subplots(len(sm.columns)-1,figsize = (12,2*len(sm.columns)-1),
                          sharex=True)
    for i,col in enumerate(sm.columns):
        if 'Quality' not in str(col):
            ax = axs[i]
            ax.axhline(processedD[col].min(),c='k',ls=':')
            ax.plot(processedD[col],label = 'without drift correction')
            ax.plot(smD[col], label = 'with drift correction')
            ax.set_ylabel(col)
            if i==0:
                ax.legend()
            ax.set_xlabel('')
    
    fig.suptitle(site)
    fig.tight_layout()
    plt.show()
        
```

```python
# function to convert vwc to a moisture depth

def total_moisture_mm(swdata1):
    swdata = swdata1.copy()
#     print(swdata.columns)
    depthcols = {str(c):c for c in swdata.columns}
#     print(depthcols)
    depths = [float(c) for c in swdata.columns]
    depths = sorted(depths)
    swdata[(depths[0])] = swdata[depthcols[str(depths[0])]]*1000*depths[0]/100
    for i in range(1,len(depths)):
        
        swdata[(depths[i])] = swdata[depthcols[str(depths[i])]]/100*(depths[i]-depths[i-1])*1000
#         print(depths[i],swdata[str(depths[i])].max())
    sm = swdata[[(d) for d in depths]]
    return sm
```

```python
# this block adds flags to all data for nonphysical moisture increases
# and saves final daily outputs to ../Data/final_sentek/
# and subdaily outputs to ../Data/temp_processing for more flagging below

# define a maximum unsaturated zone depth for each site
maxdepth = {'WP1':1.9,'WP2':2.,'GG1':6.,'GN':2, 'DD':7}

for site in sites[:]:
    
    ## Step 0: import processed temp data from the above processing block
    sm = pd.read_csv('../Data/temp_processing/vwc_processed_temp_'+site+'.csv')
    smD = pd.read_csv('../Data/temp_processing/vwc_processed_temp_'+site+'_daily.csv')
    
    for df in [sm,smD]:
        df['Date Time'] = pd.to_datetime(df['Date Time'])
        df.set_index('Date Time',inplace=True)
    ds = sm.columns[:-1]
    
    # import rain data from SILO and existing site data for comparison
    SILO = pd.read_csv('../Data/other_inputs/silo_rainfall_'+site+'.csv',
                      sep=r'\s+',header=None,)
    
    SILO.rename(columns={0:'year',
                    1:'month',
                    2:'day',
                    3:'doy',
                    4:'rain_mm'},inplace=True)
    SILO['datetime'] = pd.to_datetime(SILO.year.astype('str')+'-'+
                                     SILO.month.astype('str')+'-'+
                                     SILO.day.astype('str'))
    SILO = SILO.set_index('datetime')[['rain_mm']]
    
    try:
        weather = pd.read_csv('../Data/static_outputs/'+site+'_weather.csv')
        weather['Unnamed: 0'] = pd.to_datetime(weather['Unnamed: 0'])
        weather.set_index('Unnamed: 0',inplace=True)
        weather['Trolling'] = weather.air_temperature_C.rolling(48).mean()
        weather['date'] = weather.index.date
    except:
        print(site,'no weather')
        continue
                
    ## Step 1: convert vwc to a moisture depth for comparison to rainfall data
    gwdepth = maxdepth[site]*100

    sw = sm.copy()
    sw = sw[[c for c in ds if float(c)<gwdepth/100]]
    sw = sw.loc[weather.index.values[0]:weather.index.values[-1]].interpolate()
    ss1 = total_moisture_mm(sw)
    ss1['Quality_flags'] = sm['Quality_flags']
    
    ## Step 2: Identify dates when moisture increase far exceeds recent precipitation
    # date ranges for plotting to check cumulative mass balance
    minvs = {'GG1':'4-1-2024',
            'WP1':'4-1-2022',
            'WP2':'4-1-2022',
            'DD':'4-1-2023'}
    maxvs = {'GG1':'6-15-2026',
            'WP1':'4-1-2025',
            'WP2':'4-1-2026',
            'DD':'6-15-2026'}
    
    minv = pd.to_datetime(minvs[site])
    maxv = pd.to_datetime(maxvs[site])
    
    # plot cumulative rain from SILO, rain gage, and
    # unsaturated zone moisture increases
    fig,axs = plt.subplots(1,2,figsize = (12,5))
    ax = axs[0]
    SILO.loc[minv:maxv].plot(label='SILO',ax=ax)
    weather.rain_mm.loc[minv:maxv].resample('D').sum().plot(label='gage',ax=ax,alpha=.5)
    ss = ss1.sum(axis=1).resample('D').median().diff()
    ss = ss[ss>0].resample('D').first().to_frame()
    sss = ss.merge(SILO,left_index=True,right_index=True)
    threshnum = 5 # threshold for a moisture increase that must be corrected if above rainfall
    rollnum = 4 # window size centered around date to look for missing rainfall
    baddates = sss[(sss[0]>sss.rain_mm.rolling(rollnum,center=True).sum()*1.2)&(sss[0]>threshnum)].index.values
    sss[0] = np.where((sss[0]>sss.rain_mm.rolling(rollnum,center=True).sum()*1.2)&(sss[0]>threshnum),
                      sss.rain_mm.rolling(rollnum,center=True).max(),sss[0])
    sss[0].loc[minv:maxv].plot(label = 'sentek',marker='.',
                                                      ax=ax,alpha=.5)
    ax.legend()
    ax.set_ylabel('daily ranfall or\nmoisture increase [mm]')
    fig.suptitle(site)
    # plt.show()

    ax = axs[1]
    SILO.loc[minv:maxv].cumsum().plot(label='SILO',ax=ax)
    weather.rain_mm.loc[minv:maxv].resample('D').sum().cumsum().plot(label='gage',ax=ax,
                                                                    alpha=.5)
    sss[0].loc[minv:maxv].cumsum().plot(label = 'sentek',marker='.',
                                                               ax=ax,alpha=.5)
    ax.set_ylabel('Cumulative moisture [mm]')
    fig.tight_layout()
    plt.show()

    Icheck = sss.loc[minv:maxv].cumsum()
    irate = Icheck[0].ffill().values[-1]/Icheck['rain_mm'].values[-1]
    print(site,', interception rate: ',round(100.-irate*100,1),'%')
    
    ## Step 3: apply flags and save outputs
    fnames = ['vwc_final_',
             'vwc_final_daily_']
    for i,df in enumerate([sm,smD]):
        # flag days when moisture increases in impossible way
        baddates = [(pd.to_datetime(b).date()) for b in baddates]
        df = df.reset_index()
        df['Quality_flags'] = np.where((df['Date Time'].dt.date.isin(baddates))&(df.Quality_flags==0),
                                      3,df.Quality_flags)
        df = df.set_index('Date Time')
        if 'daily' in fnames[i]:
            loc = '../Data/final_sentek/'
        else:
            loc = '../Data/temp_processing/'
        df.to_csv(loc+fnames[i]+site+'.csv',float_format = ff)
```

```python
## add subdaily flags
for site in sites[:]:
    
    ## Step 0: import processed subdaily temp data from the above processing block
    sm = pd.read_csv('../Data/temp_processing/vwc_final_'+site+'.csv')
    
    for df in [sm,]:
        df['Date Time'] = pd.to_datetime(df['Date Time'])
        df.set_index('Date Time',inplace=True)
    ds = sm.columns[:-1]
    
    # import rain data from SILO and existing site data for comparison
    SILO = pd.read_csv('../Data/other_inputs/silo_rainfall_'+site+'.csv',
                      sep=r'\s+',header=None,)
    
    SILO.rename(columns={0:'year',
                    1:'month',
                    2:'day',
                    3:'doy',
                    4:'rain_mm'},inplace=True)
    SILO['datetime'] = pd.to_datetime(SILO.year.astype('str')+'-'+
                                     SILO.month.astype('str')+'-'+
                                     SILO.day.astype('str'))
    SILO = SILO.set_index('datetime')[['rain_mm']]
    
    try:
        weather = pd.read_csv('../Data/static_outputs/'+site+'_weather.csv')
        weather['Unnamed: 0'] = pd.to_datetime(weather['Unnamed: 0'])
        weather.set_index('Unnamed: 0',inplace=True)
        weather['Trolling'] = weather.air_temperature_C.rolling(48).mean()
        weather['date'] = weather.index.date
    except:
        print(site,'no weather')
        continue
    
    
    ## Step 1: flag days and depths 0 with no increase during daytime hours
    sm = sm.reset_index()
    days = sm['Date Time'].dt.date.drop_duplicates()
    
    baddates = {dep:[] for dep in ds}
    smdaytime = sm.set_index('Date Time').resample('H').mean().diff()
    smdaytime = smdaytime[smdaytime.index.hour.isin(list(range(6,18)))]
    incthresh = 0.0 # threshold in vwc to say that moisture is increasing
    numinc    = 3   # threshold for number of hours during day when moisture is increasing to say it's problematic
    for dep in ds[:]:
        smsub1 = smdaytime[smdaytime[dep]>incthresh]
        smsub1 = smsub1.groupby(smsub1.index.date).count()[[dep]]
        baddates[dep] = smsub1[smsub1[dep]>=numinc].index.values
        sm['Qflag_subdaily_'+dep] = np.where(sm['Date Time'].dt.date.isin(baddates[dep]),np.nan,0)
        
    ## Step 2: flag days and depths 1 with an increase but also recent rain
    ## Otherwise, if increase and no recent rain flag 2
    
    small = sm.set_index('Date Time').resample('H').mean().diff()
    rainthresh = 1.0 # threshold of rain to say it rained [mm]
    rollnum    = 4   # number of days to look around for rain
    baddates1 = {dep:[] for dep in ds}
    for dep in ds[:]:
        smsub1 = small[small[dep]<incthresh]
        smsub1 = smsub1.groupby(smsub1.index.date).count()[[dep]]
        smsub1 = smsub1.merge(SILO.rolling(rollnum,center=True).sum(),
                              left_index=True,right_index=True)
        smsub1 = smsub1[(smsub1[dep]>=numinc)&(smsub1.rain_mm<rainthresh)]
        baddates1[dep] = smsub1.reset_index()['index'].dt.date.values
        sm['Qflag_subdaily_'+dep] = np.where((sm['Date Time'].dt.date.isin(baddates1[dep]))&
                                             (sm['Qflag_subdaily_'+dep].isna()),2,
                                            sm['Qflag_subdaily_'+dep])
        sm['Qflag_subdaily_'+dep] = np.where(~(sm['Date Time'].dt.date.isin(baddates1[dep]))&
                                             (sm['Qflag_subdaily_'+dep].isna()),1,
                                            sm['Qflag_subdaily_'+dep])
        
        ## plot histogram to make sure the flags show up reasonable
        plt.hist(sm['Qflag_subdaily_'+dep],edgecolor = 'k')
        plt.ylabel('# datetimes')
        plt.title(site+': '+dep+'m')
        plt.show()
        
    
    ## Step 3: export final dataset to ../Data/final_sentek/
    sm.set_index('Date Time').to_csv('../Data/final_sentek/vwc_final_'+site+'.csv')
```
